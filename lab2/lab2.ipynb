```python
mat1 = np.random.rand(3, 3)
mat2 = np.random.rand(3, 3)
print(mat1)
print(f"\n")
print(mat2)
```

## Array operations

```python
# multiplies vector by vector
np.dot(mat1, mat2)
```

```python
# multiplies element by element
np.multiply(mat1, mat2)
```

```python
# sums the line
sum1 = np.sum(mat1, axis=0)
print(sum1)
# sums the column 
sum2 = np.sum(mat1, axis=1)
print(sum2)
```
